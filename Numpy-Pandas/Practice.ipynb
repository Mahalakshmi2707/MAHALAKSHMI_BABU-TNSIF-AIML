```python
## PROBLEM 1

import numpy as np

marks = np.array([80,85,70,75,90,95,60,65,50,55])
print("Total Marks:")
print(np.sum(marks))

print("\nAverage Marks:")
print(np.average(marks))

print("\nLowest Marks:")
print(np.array(np.min(marks)))

print("\nHighest Marks:")
print(np.array(np.max(marks)))

print("\nMarks greater than 75:")
print(marks[marks>75])

```

```python
## PROBLEM 2


import numpy as np


temperature = np.array([10,30,40,28,37,41,22])
print("Average temperature:", np.mean(temperature))


print("Highest temperature:", np.max(temperature))


print("Lowest temperature:", np.min(temperature))


print("Temperatures above 30°C:", temperature[temperature > 30])


updated_temperature = temperature + 2

print("Updated temperatures:", updated_temperature)
```

```python
## PROBLEM 4

import pandas as pd

# Product data
data = {
    "Product Name": ["Laptop", "Phone", "Headphones", "Keyboard", "Monitor"],
    "Category": ["Electronics", "Electronics", "Accessories", "Accessories", "Electronics"],
    "Price": [50000, 30000, 2000, 1500, 12000],
    "Quantity Sold": [20, 60, 80, 55, 30]
}

df = pd.DataFrame(data)


df["Total Sales"] = df["Price"] * df["Quantity Sold"]

print("Product Data:")
print(df)


highest = df.loc[df["Total Sales"].idxmax()]

print("\nProduct with highest sales:")
print(highest)


print("\nAverage product price:")
print(df["Price"].mean())


print("\nProducts with quantity sold greater than 50:")
print(df[df["Quantity Sold"] > 50])


print("\nProducts sorted by total sales:")
print(df.sort_values("Total Sales"))
```

```python
import matplotlib.pyplot as plt

# Data
months = ["January", "February", "March", "April", "May", "June"]
sales = [20000, 25000, 22000, 30000, 28000, 35000]

# Line chart
plt.plot(months, sales, marker="o", label="Sales")

plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.legend()
plt.show()


# Bar chart
plt.bar(months, sales, label="Sales")

plt.title("Monthly Sales Comparison")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt


students = ["sai", "sam", "giri", "rai", "hari", "nive", "ovi", "kala", "sana", "haley"]
marks = [45, 20, 90, 75, 75, 55, 40, 35, 60, 45]


plt.bar(students, marks)

plt.title("Student Marks")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()


excellent = 0
good = 0
average = 0
needs_improvement = 0

for mark in marks:
    if mark >= 80:
        excellent += 1
    elif mark >= 60:
        good += 1
    elif mark >= 40:
        average += 1
    else:
        needs_improvement += 1

# Values for pie chart
performance = [
    excellent,
    good,
    average,
    needs_improvement
]

labels = [
    "Excellent",
    "Good",
    "Average",
    "Needs Improvement"
]

# Pie chart
plt.pie(performance, labels=labels, autopct="%1.1f%%")

plt.title("Student Performance Distribution")
plt.show()
```
